```python
# Example of removing letters from the 'top_Mileage' column
cars_df['top_Mileage'] = cars_df['top_Mileage'].str.extract('(\d+\.?\d*)')

# Convert the cleaned column to a numeric type if needed
cars_df['top_Mileage'] = pd.to_numeric(cars_df['top_Mileage'])

# Display the DataFrame after cleaning the 'top_Mileage' column
print(cars_df[['top_Mileage']].head())

# Example of removing commas from the 'km' column
cars_df['km'] = cars_df['km'].str.replace(',', '')

# Convert the cleaned column to a numeric type if needed
cars_df['km'] = pd.to_numeric(cars_df['km'])

# Display the DataFrame after cleaning the 'km' column
print(cars_df[['km']].head())

# Function to clean and convert the price
def clean_price(price_str):
    # Remove the currency symbol
    price_str = price_str.replace('₹', '').strip()
    
    # Handle "Lakh"
    if 'Lakh' in price_str:
        price_str = price_str.replace('Lakh', '').strip()
        # Convert to numeric value (1 Lakh = 100,000)
        price_value = float(price_str.replace(',', '')) * 100000
    elif 'Crore' in price_str:
        price_str = price_str.replace('Crore', '').strip()
        # Convert to numeric value (1 Crore = 10,000,000)
        price_value = float(price_str.replace(',', '')) * 10000000
    else:
        # Just remove the commas for values in thousands
        price_value = float(price_str.replace(',', ''))
    
    return price_value

# Apply the function to the 'price' column in your DataFrame
cars_df['price'] = cars_df['price'].apply(clean_price)

# Display the updated DataFrame
print(cars_df[['price']].head())
```
